```python
# Create and display the viewer with interactive controls:
# - Automatically updates when client is acquiring (checks every 0.5s)
# - Streams at ~30 FPS when acquisition is active
# - Returns to polling mode when acquisition stops
# - Drag the corner handle to resize the widget
# - Mouse wheel to zoom in/out (zooms towards cursor position)
# - Click and drag to pan around
# - Click "Reset View" button to return to default zoom and center
# - Press 'R' key to reset the view (same as clicking Reset View button)
viewer = SimpleViewer(client=client, image=FrameType.SINGLE_FRAME_INTEGRATED)
viewer
```

```python
# NEW: Create a histogram viewer linked to the SimpleViewer
from deapi.widget import HistogramViewer

histogram = HistogramViewer(viewer=viewer)
histogram.orientation = 'vertical'  # Set histogram orientation (horizontal or vertical)
```
