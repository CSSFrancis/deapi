```python
from deapi.client import Client
from deapi.widget.simple_viewer import SimpleViewer

# Initialize your client and get an image
client = Client()  # Add connection parameters as needed
```

```python
client.connect()
```

```python
from deapi.data_types import FrameType, DataType
```

```python
# Create and display the viewer with interactive controls:
# - Automatically updates when client is acquiring (checks every 0.5s)
# - Streams at ~30 FPS when acquisition is active
# - Returns to polling mode when acquisition stops
# - Drag the corner handle to resize the widget
# - Mouse wheel to zoom in/out (zooms towards cursor position)
# - Click and drag to pan around
# - Click "Reset View" button to return to default zoom and center
# - Press 'R' key to reset the view (same as clicking Reset View button)
viewer = SimpleViewer(client=client, image=FrameType.SINGLE_FRAME_INTEGRATED)
viewer
```

```python
viewer.scalebar_location = 'top left'
```

```python
client["Image Size X (pixels)"]
```

```python
res = client.get_result()
```

```python
# You can also programmatically control the view
viewer.window_height = 256
viewer.window_width = 256
viewer.zoom = 2.0  # 2x zoom
viewer.center_x = 0.5  # Center X (0-1 normalized)
viewer.center_y = 0.5  # Center Y (0-1 normalized)

# Configure scale bar location
# Options: 'bottom right', 'bottom left', 'top right', 'top left', 'bottom', 'top', 'off'
viewer.scalebar_location = 'bottom right'  # Default

img = viewer._update_image()
```

```python
# Display the viewer
viewer
```

```python
# Test the image data
import matplotlib.pyplot as plt
plt.figure()
plt.imshow(viewer._update_image())
plt.title(f'Image: {viewer.window_width}x{viewer.window_height} pixels, Zoom: {viewer.zoom:.2f}x')
plt.colorbar()
```

```python
# Reset to default view
viewer.zoom = 1.0
viewer.center_x = 0.5
viewer.center_y = 0.5
```

```python
# Stop the viewer when done (optional, daemon thread will exit with notebook)
viewer._stop_loop = True
```
