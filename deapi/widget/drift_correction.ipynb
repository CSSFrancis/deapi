"In Situ Experiment with Drift Correction
========================================

This notebook demonstrates how to perform drift correction on
an insitu using the deapi, panel and the DENS API.


```python
import deapi
from deapi import Client, FrameType
import panel as pn
import ipywidgets as ipw
pn.extension("ipywidgets")
```

```python
client = Client()
client.connect()
```

```python
# Combine into a column layout
scan_row = pn.Row(client["Scan - Size X"].widget,
                  client["Scan - Size Y"].widget,
                  )
rep_row = pn.Row(client["Scan - Repeats"].widget, client["Scan - Camera Frames Per Point"].widget)
start_button = client.start_acquisition.widget
start_button.button.width = 200
scan_params = pn.Column(
    "### Scan Parameters",
    scan_row,
    rep_row,
    start_button,
width =215,)

# Add viewers
diffraction_image = client.get_live_result(frame_type=FrameType.SINGLE_FRAME_INTEGRATED)
virtual_image = client.get_live_result(frame_type=FrameType.VIRTUAL_IMAGE0)

stem_viewer = pn.Row(scan_params,diffraction_image, virtual_image, sizing_mode='stretch_width')
```

```python
date   = ipw.DatePicker(description='Date')
slider = ipw.FloatSlider(description='Float')
play   = ipw.Play()

layout = ipw.HBox(children=[date, slider, play])

pn = pn.panel(layout)
```

```python
pn
```

```python
def panel_viewer(result_viewer):
    """Wrap ResultViewer for Panel compatibility"""
    import ipywidgets as widgets

    # Wrap in an ipywidgets Output to force rendering
    output = widgets.Output()
    with output:
        display(result_viewer)

    return pn.pane.IPyWidget(output)

# Use it:
pane = panel_viewer(diffraction_image)
```

```python
# Option 1: Use ipywidgets.Output to wrap anywidget
import ipywidgets as widgets

diff_output = widgets.Output()
with diff_output:
    display(diffraction_image)

virt_output = widgets.Output()
with virt_output:
    display(virtual_image)

# Now wrap the Output widgets
stem_viewer = pn.Row(
    scan_params,
    pn.pane.IPyWidget(diff_output, width=600, height=600),
    pn.pane.IPyWidget(virt_output, width=600, height=600),
    sizing_mode='stretch_width'
)
```
